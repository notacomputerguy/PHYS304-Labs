```python
import numpy as np
import matplotlib.pyplot as plt
import os

nsteps = 30


def euler_newton(force, times, x0, v0, m):
    v = np.zeros(nsteps)
    x = np.zeros(nsteps)
    x[0]=x0
    v[0]=v0
    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v[i+1] = dt*a + v[i]
        x[i+1] = v[i]*dt + x[i]
    return x, v
```
